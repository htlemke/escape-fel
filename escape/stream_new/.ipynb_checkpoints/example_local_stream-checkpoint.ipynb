# escape stream_new — local test stream example

This notebook exercises `escape.stream_new` against a **synthetic local bsread stream**,
so it works without any beamline connection.

The test stream broadcasts pump-probe channels at ~100 Hz:

| Channel | Description |
|---------|-------------|
| `i0` | Incoming photon flux (shot noise) |
| `i` | Transmitted / diffracted signal |
| `t` | Pump-probe delay (ps) |
| `pump_on` | Laser on (1) / off (0) flag |
| `i_pump` | Pump perturbation amplitude |
| `drift` | Slow beam drift |

**Prerequisites:** `%matplotlib widget` (ipympl) for live updating plots.

```python
%matplotlib widget
```

## 1 — Imports

```python
import sys, time
sys.path.insert(0, '/das/home/lemke_h/mypy/escape-fel/')

import numpy as np
import matplotlib.pyplot as plt

from escape.stream_new import (
    EventWorker, EscData, EventSource, StreamContext,
    LocalEventHandler, TestStream,
)
```

## 2 — Start the synthetic test stream

`TestStream` launches `createStream()` in a child process, binding a bsread Sender
to `localhost:9999`.  It broadcasts all channels continuously until stopped.

```python
ts = TestStream(port=9999, interval=0.01)   # ~100 Hz
ts.start()
time.sleep(0.5)   # give the sender a moment to bind
ts
```

## 3 — Create an EventWorker and EscData objects

`LocalEventHandler` connects directly to `localhost:9999` with `all_channels=True`
(no dispatcher needed).  `EventWorker` runs the receive loop in a daemon thread.

```python
ew = EventWorker(eventHandler=LocalEventHandler(), make_default=True)

i0      = EscData(source=EventSource('i0',      ew, unit='a.u.'))
i       = EscData(source=EventSource('i',       ew, unit='a.u.'))
t       = EscData(source=EventSource('t',       ew, unit='ps'))
pump_on = EscData(source=EventSource('pump_on', ew, unit='bool'))
drift   = EscData(source=EventSource('drift',   ew, unit='a.u.'))
```

## 4 — Start accumulation

Calling `accumulate(True)` registers the channel with the EventWorker and starts
collecting events into an in-memory circular buffer (default 1000 events per step).

```python
i0.accumulate(True)
i.accumulate(True)
t.accumulate(True)
pump_on.accumulate(True)

time.sleep(2)   # collect a couple of hundred events

print(f"i0 events collected : {len(i0)}")
print(f"i  events collected : {len(i)}")
print(f"t  events collected : {len(t)}")
```

## 5 — Inspect raw buffer contents

```python
# data[0] is a deque of values for scan step 0 (no-scan = single step)
i0_arr = np.array(i0.data[0])
print(f"i0: mean={i0_arr.mean():.4f}  std={i0_arr.std():.4f}  n={len(i0_arr)}")

i_arr = np.array(i.data[0])
print(f"i:  mean={i_arr.mean():.4f}  std={i_arr.std():.4f}")
```

## 6 — Live histogram

`plot_hist()` opens a figure and starts a background thread that refreshes every
0.5 s.  **Closing the figure automatically stops accumulation for `i0`.**

```python
fig_h, ax_h = plt.subplots(figsize=(6, 3))
fig_h.suptitle('i0 — event count histogram')

# plot_hist returns the HistPlot object so you can stop it later
hp = i0.plot_hist(update=0.5, axes=ax_h)
```

## 7 — Live scatter / correlation plot

Shows the last 400 events of `i` vs `i0`, matched by pulse ID.

```python
fig_c, ax_c = plt.subplots(figsize=(5, 5))
cp = i.plot_corr(i0, Npoints=400, update=0.5, axes=ax_c)
```

## 8 — StreamContext: managed acquisition lifetime

Use `StreamContext` as a context manager to accumulate for a fixed duration,
then stop automatically — no need to call `accumulate(False)` manually.

You can also tie a context to a figure window so that closing the figure
stops acquisition, and re-calling the block restarts it.

```python
# Accumulate drift for exactly 3 seconds, then stop.
with StreamContext(drift):
    print('Accumulating drift for 3 s…')
    time.sleep(3)

drift_arr = np.array(drift.data[0])
print(f'Collected {len(drift_arr)} drift events,  mean = {drift_arr.mean():.4f}')
```

```python
# Tie a StreamContext to a figure: accumulation stops when the figure is closed.
fig_d, ax_d = plt.subplots(figsize=(6, 3))
fig_d.suptitle('drift  (close window to stop)')

ctx = StreamContext(drift)
ctx.tie_to_figure(fig_d)   # wires the close_event
ctx.start()

mp = drift.plot_med(update=0.5, axes=ax_d)
print('Close the figure above to automatically stop drift acquisition.')
```

## 9 — per-object `.acquire()` shorthand

`escdata.acquire()` returns a `StreamContext` for that single object.

```python
with i0.acquire():
    time.sleep(2)

print(f'i0 events after acquire block: {len(i0)}')
```

## 10 — Stop everything

```python
# Stop all live-plot update threads
for plot_obj in [hp, cp, mp]:
    plot_obj.stop()

# Stop accumulation for all channels
for ch in [i0, i, t, pump_on, drift]:
    ch.accumulate(False)

# Shut down the event loop
ew.stopEventLoop()

# Kill the test stream process
ts.stop()

print('Done.')
```
