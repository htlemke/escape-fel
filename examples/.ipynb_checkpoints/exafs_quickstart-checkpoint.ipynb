# EXAFS Quickstart

This notebook runs the full `escape.exafs` pipeline — normalize, extract chi(k),
Fourier transform to chi(R) — on **real experimental data**: Cu K-edge EXAFS of a
copper metal foil at room temperature (APS beamline 13ID, via the
[xraylarch](https://github.com/xraypy/xraylarch) example data set).

If you're brand new to EXAFS, read `README.md` first for the physics background.
For a step-by-step "build it yourself" version of this same pipeline, see
`02_understanding_the_steps.ipynb`.


```python
import os
import escape.exafs, os  # data path resolved from the installed package

import numpy as np
import matplotlib.pyplot as plt
from escape import exafs as ex

print("escape.exafs version:", ex.__version__)
```

## 1. Load the data

Any simple 2-column "energy, mu" text file works. Our demo file has 3 columns
(energy, mu, I0); we just tell `read_columns` which ones to use.


```python
energy, mu = ex.read_columns(os.path.join(os.path.dirname(escape.exafs.__file__), "data", "cu_rt01.xmu"), energy_col=0, mu_col=1)

print(f"{len(energy)} data points, {energy.min():.1f} to {energy.max():.1f} eV")

fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(energy, mu)
ax.set_xlabel("Energy (eV)")
ax.set_ylabel(r"$\mu(E)$ (raw)")
ax.set_title("Raw Cu foil XAS spectrum")
plt.show()
```

## 2. Normalize: pre-edge subtraction + edge-step normalization

`pre_edge()` does three things automatically (you can override any of them):

- estimates **E0** (the edge energy) as the peak of dmu/dE,
- fits a straight line through the **pre-edge** region and subtracts it,
- fits a smooth polynomial through the **post-edge** region to get the
  **edge step**, and divides by it.


```python
pre = ex.pre_edge(energy, mu)

print(f"E0          = {pre.e0:.2f} eV")
print(f"edge step   = {pre.edge_step:.4f}")

fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(pre.energy, pre.mu, label="raw mu(E)")
ax.plot(pre.energy, pre.pre_edge, "--", label="pre-edge line")
ax.plot(pre.energy, pre.post_edge, "--", label="post-edge polynomial")
ax.axvline(pre.e0, color="gray", lw=0.8, ls=":", label="E0")
ax.set_xlabel("Energy (eV)")
ax.set_ylabel(r"$\mu(E)$")
ax.legend()
ax.set_title("Pre-edge / post-edge fits")
plt.show()
```

## 3. Extract chi(k): remove the smooth background

`autobk()` fits a smooth cubic spline mu0(k) through the post-edge data and
subtracts it, leaving the oscillatory EXAFS signal chi(k). The key physical
input is **`rbkg`**: "I don't believe there's a real atomic shell closer than
this many Angstroms" (for a metal foil like Cu, 1.0 Ang is a safe, standard
choice — the true first shell is at 2.55 Ang).


```python
bkg = ex.autobk(energy, mu, pre.e0, pre.edge_step, rbkg=1.0)

ex.plot_bkg(pre, bkg, emin=-50, emax=500)
plt.show()
```

```python
ex.plot_chik(bkg.k, bkg.chi, kweight=2)
plt.show()
```

## 4. Fourier transform: chi(k) -> chi(R)

The oscillations in chi(k) are (approximately) a sum of damped sine waves, one
per coordination shell, at frequency `2 * R`. Fourier transforming a
k-weighted, windowed chi(k) turns those into peaks vs. distance.

- `kweight` (usually 1, 2, or 3) emphasizes different parts of the k-range —
  higher kweight emphasizes heavier backscatterers / high-k structure.
- `window` tapers chi(k) smoothly to zero at `kmin`/`kmax` so you don't get
  spurious ripples from an abrupt cutoff.


```python
r, chir, kgrid = ex.ft_windowed(
    bkg.k, bkg.chi, kweight=2, window="hanning", kmin=3, kmax=13, dk=1.0
)

ex.plot_chir(r, chir, rmax=6)
plt.show()
```

## 5. Reading the result

The big peak near **R ≈ 2.2-2.3 Å** is the Cu-Cu nearest-neighbor shell.
Note that this is **not** the true bond distance (2.55 Å in metallic
copper) — EXAFS peaks are shifted ~0.3-0.5 Å below the real distance by a
scattering phase shift that depends on the atom type. Getting the *true*
distance requires fitting chi(k) or chi(R) against a theoretical standard
(e.g. from FEFF), which is beyond the scope of this package — but everything
you need to feed into such a fit (chi(k), the k-grid, edge step, E0) is
sitting in the `bkg` object above.

The smaller peak(s) around 4-5 Å are real too: in a close-packed metal like
copper, "focused" multiple-scattering paths through collinear atoms produce
strong extra peaks beyond the simple single-scattering shells.


```python
print("chi(k) is available as bkg.k, bkg.chi")
print("You could pass these into your own least-squares fit against a FEFF")
print("calculation (e.g. using lmfit) to get quantitative distances, ")
print("coordination numbers, and Debye-Waller factors.")
```
